```python
import os
import shutil

path_to_validation = "val_split_images.txt"

path_to_train_source = "/home/pablo.canosa/ssd/datasets_pablo/rios_FBP_BALANCED_REFERENCED_2026_validation/img_train"
path_to_mask_source = "/home/pablo.canosa/ssd/datasets_pablo/rios_FBP_BALANCED_REFERENCED_2026_validation/gt_train"

path_to_val_dest = "/home/pablo.canosa/ssd/datasets_pablo/rios_FBP_BALANCED_REFERENCED_2026_validation/img_val"
path_to_mask_dest = "/home/pablo.canosa/ssd/datasets_pablo/rios_FBP_BALANCED_REFERENCED_2026_validation/gt_val"


os.makedirs(path_to_val_dest, exist_ok=True)
os.makedirs(path_to_mask_dest, exist_ok=True)


with open(path_to_validation, 'r') as f:
    val_images =[line.strip() for line in f.readlines() if line.strip()]

moved_count = 0

for img_filename in val_images:
    base_name = os.path.splitext(img_filename)[0]
    
    
    mask_filename = f"{base_name}.png"


    img_src = os.path.join(path_to_train_source, img_filename)
    img_dst = os.path.join(path_to_val_dest, img_filename)


    mask_src = os.path.join(path_to_mask_source, mask_filename)
    mask_dst = os.path.join(path_to_mask_dest, mask_filename)


    if os.path.exists(img_src):
        shutil.move(img_src, img_dst)
    else:
        print(f"Warning: Image not found -> {img_src}")


    if os.path.exists(mask_src):
        shutil.move(mask_src, mask_dst)
    else:
        print(f"Warning: Mask not found -> {mask_src}")

    moved_count += 1

print(f"Successfully moved {moved_count} image/mask pairs to the validation folders.")
```
