```python
import os
import numpy as np
from PIL import Image
from tqdm import tqdm

NUM_CLASSES = 25

train_folder = "/home/pablo.canosa/ssd/datasets_pablo/cutGID4BV2/annotations/training"
test_folder = "/home/pablo.canosa/ssd/datasets_pablo/cutGID4BV2/annotations/validation"



def count_pixels(folder):
    counts = np.zeros(NUM_CLASSES, dtype=np.int64)

    files = [f for f in os.listdir(folder) if f.endswith((".png", ".tif", ".jpg"))]

    for f in tqdm(files, desc=f"Processing {folder}"):
        path = os.path.join(folder, f)

        mask = np.array(Image.open(path)).astype(np.int64)

        hist = np.bincount(mask.flatten(), minlength=NUM_CLASSES)
        counts += hist

    return counts


# Count pixels
train_counts = count_pixels(train_folder)
test_counts = count_pixels(test_folder)

total_counts = train_counts + test_counts

# Remove unlabeled pixels (class 0) from totals
train_total = train_counts[1:].sum()
test_total = test_counts[1:].sum()
total_pixels = total_counts[1:].sum()

# Percentages (ignoring class 0)
train_pct = np.zeros(NUM_CLASSES)
test_pct = np.zeros(NUM_CLASSES)
total_pct = np.zeros(NUM_CLASSES)

train_pct[1:] = 100 * train_counts[1:] / train_total
test_pct[1:] = 100 * test_counts[1:] / test_total
total_pct[1:] = 100 * total_counts[1:] / total_pixels


# Print results
print("\nClass distribution (%) (ignoring class 0 - unlabeled)")
print(f"{'Class':>5} {'Train':>10} {'Test':>10} {'Total':>10}")
print("-"*42)

for i in range(1, NUM_CLASSES):
    print(f"{i:>5} {train_pct[i]:>10.3f} {test_pct[i]:>10.3f} {total_pct[i]:>10.3f}")
```

```python
def read_rawPV(filename):
	# Lectura de imaxe Pixel-Vector en RAW
	imaxe = np.fromfile(filename, dtype='int32')
	B = imaxe[0]
	H = imaxe[1]
	V = imaxe[2]
	#Sacamos os datos da cabeceira e pasamos a 3D o array
	imaxe = imaxe[3:].reshape((V, H, B))

	return imaxe

from osgeo import gdal
import cv2
def read_rawb_NirRGB(file_name):
    # check if the file has tif or rawb extension
    if file_name.endswith('.rawb'): 
        with open(file_name, "rb") as f:
            sizes = np.fromfile(f, dtype=np.int32, count=3)
            image_data = np.fromfile(f, dtype=np.int32)
            image = image_data.reshape(sizes)
            image = image.astype(np.float32)
    elif file_name.endswith('.tif'):
        image = gdal.Open(file_name).ReadAsArray()
        image = np.asarray(image, dtype=np.float32)
        # shape to (H, W, C)
        image = np.moveaxis(image, 0, -1)
    elif file_name.endswith('.raw'):# para las imágenes de rios
        image = read_rawPV(file_name)
        image = np.asarray(image, dtype=np.float32)
    elif file_name.endswith('.jpg'):
        image =  cv2.imread(file_name)
        image = np.asarray(image, dtype=np.float32)
        
    return image

def read_rawb_RGB(file_name,version):
    img = read_rawb_NirRGB(file_name)

    if(version == "gaofen"):
        return img[:, :, [2,1,0]]  # Delete Nir channel (B-G-R-NIR) -> (R-G-B)
    elif(version == "rios"):
        return img[:, :, [2,1,0]]  # Las imagenes de rios vienen en B-G-R-RedEdge-NIR
    elif(version == "LADOS"):
        return img[:, :, [0,1,2]]  # Las imagenes de LADOS ya tienen tres bandas, pero asi nos aseguramos de que no haya incompatibilidades al llamar al metodo
    elif(version == "LANDCOVERAI"):
        return img[:, :, [0,1,2]]  # Las imagenes de LANDCOVERAI ya tienen tres bandas, pero asi nos aseguramos de que no haya incompatibilidades al llamar al metodo
    elif(version == "rios_fusion"):
        # I receive the image in B-G-R-RedEdge-NIR, for testing purposes I want to return RGB-RedEdge-NIR-NDVI, so I will calculate the NDVI and return the image in the format RGB-RedEdge-NIR-NDVI
        # Calculate NDVI
        nir = img[:, :, 4]
        red = img[:, :, 2]
        ndvi = (nir - red) / (nir + red + 1e-6)
        # Return RGB-RedEdge-NIR-NDVI
        return np.concatenate((img[:, :, [2,1,0]], img[:, :, 3:5], ndvi[:, :, np.newaxis]), axis=2)
    elif(version == "gaofen_fusion"):
        # I receive the image in B-G-R-NIR
        blue = img[:, :, 0]
        green = img[:, :, 1]
        red = img[:, :, 2]
        nir = img[:, :, 3] 

        ndvi = (nir - red) / (nir + red + 1e-6)

        ndwi = (green - nir) / (green + nir + 1e-6)

        rgb_bands = img[:, :, [2, 1, 0]]  # BGR a RGB

        nir_expanded = nir[:, :, np.newaxis]
        ndvi_expanded = ndvi[:, :, np.newaxis]
        ndwi_expanded = ndwi[:, :, np.newaxis]

        return np.concatenate((rgb_bands, nir_expanded, ndvi_expanded, ndwi_expanded), axis=2)

    else: # Throw error
        raise ValueError("Version not recognized. Use 'gaofen', 'rios', 'LADOS' or 'LANDCOVERAI'.")
```

```python
import os
import numpy as np
from tqdm import tqdm
import matplotlib.pyplot as plt

path_train = "/home/pablo.canosa/ssd/datasets_pablo/cutGID4BV2/images/training"

# 1. Recopilamos todos los archivos válidos primero
files =[f for f in os.listdir(path_train) if f.endswith((".rawb", ".tif", ".raw", ".jpg"))]

# Inicializamos las sumas en float64 para evitar desbordamientos de memoria
# Asumimos 6 bandas para "gaofen_fusion" (ej. RGB + NIR + NDVI + NDWI)
num_bands = 6
sum_bands = np.zeros(num_bands, dtype=np.float64)
sum_sq_bands = np.zeros(num_bands, dtype=np.float64)
count_pixels = 0

# 2. Bucle para calcular la media y la desviación típica real
for f in tqdm(files, desc="Reading training images"):
    img_path = os.path.join(path_train, f)
    # Leemos la imagen con tu función
    img = read_rawb_RGB(img_path, version="gaofen_fusion")
    
    # Acumulamos la suma de los valores y de los valores al cuadrado
    sum_bands += img.sum(axis=(0, 1))
    sum_sq_bands += (img ** 2).sum(axis=(0, 1))
    
    # Acumulamos el número total de píxeles procesados
    count_pixels += img.shape[0] * img.shape[1]

# 3. Cálculos matemáticos finales
mean = sum_bands / count_pixels
variance = (sum_sq_bands / count_pixels) - (mean ** 2)
std = np.sqrt(variance)

print("\n--- RESULTADOS ESTADÍSTICOS ---")
print(f"PIXEL_MEAN (6 bands): {np.round(mean, 4).tolist()}")
print(f"PIXEL_STD (6 bands):  {np.round(std, 4).tolist()}\n")
        
```

```python
# 4. Mostrar una imagen aleatoria en RGB
print("Generando visualización de imagen aleatoria...")
random_file = np.random.choice(files)
random_img_path = os.path.join(path_train, random_file)
random_img = read_rawb_RGB(random_img_path, version="gaofen_fusion")

# Extraemos solo las bandas RGB (índices 0, 1, 2)
rgb_img = random_img[:, :, :3].copy()

# Normalización robusta para visualización (ignora outliers solares/sombras extremas)
p2, p98 = np.percentile(rgb_img, (2, 98))

if p98 > p2:  # Evitar división por cero si la imagen es plana
    rgb_normalized = (rgb_img - p2) / (p98 - p2)
else:
    rgb_normalized = rgb_img

# Clipiamos entre 0 y 1 para que Matplotlib lo entienda perfectamente
rgb_normalized = np.clip(rgb_normalized, 0, 1)

# Ploteamos
plt.figure(figsize=(8, 8))
plt.imshow(rgb_normalized)
plt.title(f"Random Training Image (RGB)\n{random_file}")
plt.axis('off')
plt.show()
```
